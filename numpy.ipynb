```python
array=np.array([3,4.5,4+5j,0])
real=np.isreal(array)
array_real=array[real]
print(array_real)
print(real)
print('-'*30)
imag=np.iscomplex(array)
array_imag=array[imag]
print(array_imag)
print(imag)
```

```python
##buy and sell stocks
price=[[40,35,20],[21,48,70]]
prices=np.array(price)
print(prices)
prices=prices.flatten()
buying_price=np.min(prices)
index=list(prices).index(buying_price)
new_prices=prices[index:]
selling_prices=np.max(new_prices)
profit=selling_prices-buying_price
print(profit)
```

```python
np.random.seed(21)

# Code starts here

# create random 5x5 array
Z = np.random.random((5,5))

# minimum and maximum values
Zmax, Zmin = Z.max(), Z.min()

# normalize
Z_norm = (Z - Zmin)/(Zmax - Zmin)

# display
print(Z_norm)

# Code ends here
```
